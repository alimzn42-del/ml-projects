```python
import numpy as np

rng = np.random.default_rng(0)
X = rng.uniform(0, 10, 200)
y = 2.5 * X + 4 + rng.normal(0, 0.5, 200)

w, b, lr = 0.0, 0.0, 0.01
hist = []

for epoch in range(1100):
    yh = w * X + b
    L  = np.mean((y - yh) ** 2)

    dw = np.mean(-2 * (y - yh) * X)
    db = np.mean(-2 * (y - yh))

    w -= lr * dw
    b -= lr * db
    hist.append(L)

    if epoch % 40 == 0:
        print(f"epoch {epoch:>3} | L={L:8.4f} | w={w:6.3f} | b={b:6.3f}")

print(f"\nfinal: w={w:.3f}  b={b:.3f}   (true: 2.5, 4.0)")
```

```python
def loss(w, b):
    return np.mean((y - (w * X + b)) ** 2)

w_t, b_t = 1.7, 0.9          # نقطة اختبار عشوائية
eps = 1e-5

yh = w_t * X + b_t
dw_a = np.mean(-2 * (y - yh) )
db_a = np.mean(-2 * (y - yh))

dw_n = (loss(w_t + eps, b_t) - loss(w_t - eps, b_t)) / (2 * eps)
db_n = (loss(w_t, b_t + eps) - loss(w_t, b_t - eps)) / (2 * eps)

for nm, a, n in [("dw", dw_a, dw_n), ("db", db_a, db_n)]:
    rel = abs(a - n) / (abs(a) + abs(n))
    print(f"{nm} | analytic={a:12.6f} | numeric={n:12.6f} | rel={rel:.2e}")
```
